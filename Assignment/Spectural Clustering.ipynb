```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
dataset
```

```python
X = dataset.iloc[:, 3:5].values
```

```python
X
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import SpectralClustering
from sklearn.metrics import silhouette_score
list1 = []
for k in range(2, 11):
    model = SpectralClustering(n_clusters=k,random_state=0)
    labels = model.fit_predict(X_scaled) 
    score = silhouette_score(X_scaled, labels)
    list1.append(score)
    print("Clusters:", k, "Silhouette Score:", score)
plt.plot(range(2, 11), list1, marker='o')
plt.title('Silhouette Method')
plt.xlabel('Number of clusters')
plt.ylabel('Silhouette Score')
plt.show()
```

```python
from sklearn.cluster import SpectralClustering
spectural = SpectralClustering(n_clusters = 6, random_state = 42)
y_spectural = spectural.fit_predict(X)
```

```python
supervised=dataset
```

```python
supervised
```

```python
supervised['Cluster_group']=y_spectural
```

```python
supervised
```

```python
supervised.to_csv("cluster.csv",index=False)
```

```python
import seaborn as sns
facet = sns.lmplot(data=supervised, x=supervised.columns[3], y=supervised.columns[4], hue=supervised.columns[5], 
                   fit_reg=False, legend=True, legend_out=True)
```
